# Charting in Colaboratory

A common use for notebooks is data visualization using charts. Colaboratory makes this easy with several charting tools available as Python imports.

```
import pandas as pd
import zipfile
import os

zip_file_path = '/content/bank+marketing (2).zip'
extract_dir = '/content/extracted_bank_data'
os.makedirs(extract_dir, exist_ok=True)

# Unzip the main file, which contains other zip files
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

# The inner zip file is 'bank.zip'
inner_zip_path = os.path.join(extract_dir, 'bank.zip')

# Create a new directory to extract the contents of bank.zip
inner_extract_dir = os.path.join(extract_dir, 'bank_data_csv')
os.makedirs(inner_extract_dir, exist_ok=True)

# Extract bank.zip
with zipfile.ZipFile(inner_zip_path, 'r') as zip_ref:
    zip_ref.extractall(inner_extract_dir)

# Now read the specific 'bank.csv' file from the extracted directory
csv_file_path = os.path.join(inner_extract_dir, 'bank.csv')
df = pd.read_csv(csv_file_path, encoding='utf-8', sep=';')
```

```
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.countplot(x='job', data=df)
plt.xticks(rotation=45)
plt.title("Job Distribution")
plt.show()
```

2. Marital Status (Bar Chart)

```
plt.figure(figsize=(6,4))
sns.countplot(x='marital', data=df)
plt.title("Marital Status")
plt.show()
```

3. Target Variable (Pie Chart)

```
df['y'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(5,5)
)
plt.title("Subscription (Yes/No)")
plt.ylabel("")
plt.show()
```

4. Housing Loan (Pie Chart)

```
df['housing'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(5,5)
)
plt.title("Housing Loan")
plt.ylabel("")
plt.show()
```

5. Age Distribution (Histogram)

```
plt.figure(figsize=(8,5))
plt.hist(df['age'], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

6. Balance Distribution (Histogram)

```
plt.figure(figsize=(8,5))
plt.hist(df['balance'], bins=30)
plt.title("Balance Distribution")
plt.xlabel("Balance")
plt.ylabel("Frequency")
plt.show()
```

7. Balance Outliers (Box Plot)

```
plt.figure(figsize=(8,4))
sns.boxplot(x=df['balance'])
plt.title("Balance Box Plot")
plt.show()
```
